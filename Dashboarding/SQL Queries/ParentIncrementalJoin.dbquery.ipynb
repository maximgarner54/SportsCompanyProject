```sql
COPY INTO sportsproject.gold.fact_orders
FROM (
  SELECT 
    CAST(date AS DATE) AS date,
    product_code,
    customer_code,
    CAST(sold_quantity AS BIGINT) AS sold_quantity
  FROM '/Volumes/sportsproject/gold/parent_incremental_update'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true');
```
